```python
import requests
import json
import pandas as pd

from fraud_detection_pipeline.utils.constants import (
    PARENT_DIR,
    PROCESSED_DIRECTORY,
)
```

```python
df = pd.read_csv(PARENT_DIR / PROCESSED_DIRECTORY / "testdata.csv")
data = df[df["is_fraud"] == 1].iloc[:5].drop(columns=["is_fraud"]).values.tolist()
data
```

```python
with requests.get("http://localhost:5033/") as response:
    print(response.text)
```

```python
url = "http://localhost:5033/predict"
headers = {"Content-Type": "application/json"}
payload = {"inputs": data}

response = requests.post(url, headers=headers, data=json.dumps(payload))
response.raise_for_status()
predictions = response.json()
print(predictions)
```
